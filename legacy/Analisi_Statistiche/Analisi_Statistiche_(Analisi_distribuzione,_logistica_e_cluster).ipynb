````python
def calcola_percentuali(data, variabile):
    """
    Calcola le percentuali per una variabile rispetto a 'BEV dummy' in un DataFrame.

    :param data: DataFrame contenente i dati.
    :param variabile: Il nome della variabile categoriale per cui calcolare le percentuali (string).
    :return: DataFrame con le percentuali calcolate.
    """
    counts = data.groupby(['BEV dummy', variabile]).size().unstack(fill_value=0)     # Calcolare i conteggi per le combinazioni di 'BEV dummy' e la variabile
    totali = counts.sum(axis=0)     # Calcolare il totale per ciascun livello della variabile
    percentuali = counts.div(totali, axis=1) * 100    # Calcolare le percentuali
    percentuali = percentuali.stack().reset_index(name='Percentuale')     # Trasformare in formato "long" e aggiungere la colonna 'Percentuale'

    return percentuali

percentuali_gen = calcola_percentuali(data_originale, 'Genere')
percentuali_eta = calcola_percentuali(data_originale, 'Classe d\'età')
percentuali_reddito = calcola_percentuali(data_originale, 'Classe Reddito Familiare')
percentuali_istruzione = calcola_percentuali(data_originale, 'Livello di istruzione')
percentuali_casa_proprieta = calcola_percentuali(data_originale, 'Casa di proprietà')
percentuali_casa_indipendente = calcola_percentuali(data_originale, 'Casa Indipendente')
percentuali_auto_precedente = calcola_percentuali(data_originale, 'Tipologia di auto precedente')
````

````python
def create_bins(data, col_name, bins, suffix):
    bin_col_name = f'{col_name}_bin'
    bin_num_col_name = f'{col_name}_bin_num'
    data[bin_col_name] = pd.cut(data[col_name], bins=bins, right=False)     # Aggiungi la colonna dei bin
    data[bin_num_col_name] = data[bin_col_name].apply(lambda x: x.left)     # Aggiungi la colonna con il numero del bin (valore inferiore di ciascun intervallo)
    return data

bins_viaggi = np.arange(0, 1050, 50)
bins_nviaggi = np.arange(0, 12, 1)
bins_importanza = np.arange(-3, 3.5, 0.5)
bins_dist = np.arange(0, 60, 5)
bins_vmt = np.arange(0, 51000, 1000)

data_biva = create_bins(data_biva, 'Importanza di ridurre le emissioni di gas serra', bins_importanza, 'Importanza')
data_biva = create_bins(data_biva, 'Viaggio più lungo negli ultimi 12 mesi', bins_viaggi, 'viaggi')
data_biva = create_bins(data_biva, 'Numero di viaggi superiori a 200 miglia negli ultimi 12 mesi', bins_nviaggi, 'nviaggi')
data_biva = create_bins(data_biva, 'Distanza casa-lavoro', bins_dist, 'dist')
data_biva = create_bins(data_biva, 'VMT annuo', bins_vmt, 'vmt')

data_biva.head()
````

### 5.1.2. Analisi VIF

````python
vif_data = pd.DataFrame()
vif_data["feature"] = Xind.columns
vif_data["VIF"] = [variance_inflation_factor(Xind.values, i) for i in range(len(Xind.columns))]
print(vif_data)
````

````python
logit_model = sm.Logit(y, Xind)
result = logit_model.fit()
ll_null = logit_model.loglikeobs(np.zeros(logit_model.exog.shape[1])).sum()
ll_model = result.llf
mcfadden_rsquared = 1 - (ll_model / ll_null)
print(f"McFadden's R-squared: {mcfadden_rsquared}")
````

````python
Xind_ridotto = Xind.drop(["Numero persone in famiglia", "Casa di proprietà"], axis=1)

vif_data_ridotto = pd.DataFrame()
vif_data_ridotto["feature"] = Xind_ridotto.columns
vif_data_ridotto["VIF"] = [variance_inflation_factor(Xind_ridotto.values, i) for i in range(len(Xind_ridotto.columns))]
print(vif_data_ridotto)
````

````python
def sono_simili(val1, val2, tolleranza=0.0001):
    return abs(val1 - val2) < tolleranza

# Modello completo
logit_model_vif = sm.Logit(y, Xind)
result_vif = logit_model_vif.fit()
ll_null_vif = logit_model_vif.loglikeobs(np.zeros(logit_model_vif.exog.shape[1])).sum()
ll_model_vif = result_vif.llf
mcfadden_rsquared_vif = 1 - (ll_model_vif / ll_null_vif)
print(f"\nR2 di McFadden per modello completo: {mcfadden_rsquared_vif:.6f}")

# Modello ridotto
logit_model_vif2 = sm.Logit(y, Xind_ridotto)
result_vif2 = logit_model_vif2.fit()
ll_null_vif2 = logit_model_vif2.loglikeobs(np.zeros(logit_model_vif2.exog.shape[1])).sum()
ll_model_vif2 = result_vif2.llf
mcfadden_rsquared_vif2 = 1 - (ll_model_vif2 / ll_null_vif2)
print(f"\nR2 di McFadden per modello ridotto: {mcfadden_rsquared_vif2:.6f}")
print("\n===============================================================")
# Confronto
if sono_simili(mcfadden_rsquared_vif, mcfadden_rsquared_vif2):
    print("|    Entrambi i modelli hanno un R2 simile (entro 4 decimali).")
    print("|    Si può procedere con il modello ridotto.")
elif mcfadden_rsquared_vif > mcfadden_rsquared_vif2:
    print("Il modello completo ha un R2 maggiore di quello ridotto.")
else:
    print("Il modello ridotto ha un R2 maggiore di quello completo.")
print("===============================================================")
````
